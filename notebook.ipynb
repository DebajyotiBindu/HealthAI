```python
# ============================================================
# LUNG CANCER DETECTION - COMPLETE MACHINE LEARNING PIPELINE
# Dataset:
# jillanisofttech/lung-cancer-detection
#
# Models:
# 1. Random Forest
# 2. Logistic Regression
# 3. SVC
# 4. KNN
# 5. ANN
#
# Evaluation:
# Accuracy, Precision, Recall, ROC-AUC
#
# Additional:
# - Data cleaning
# - Feature engineering
# - Class balance analysis
# - Training-only oversampling
# - Confusion matrices
# - Comparative graphs
# - Best-model selection
# - Pickle
# ============================================================


# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

!pip -q install kaggle imbalanced-learn


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import os
import glob
import pickle
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestClassifier

from sklearn.linear_model import LogisticRegression

from sklearn.svm import SVC

from sklearn.neighbors import KNeighborsClassifier

from sklearn.neural_network import MLPClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    roc_auc_score,
    confusion_matrix
)

from imblearn.over_sampling import RandomOverSampler
from imblearn.pipeline import Pipeline as ImbPipeline

warnings.filterwarnings("ignore")

RANDOM_STATE = 42


# ============================================================
# 3. KAGGLE AUTHENTICATION
# ============================================================

print("=" * 70)
print("KAGGLE AUTHENTICATION")
print("=" * 70)

# Upload kaggle.json
uploaded = files.upload()

os.makedirs("/root/.kaggle", exist_ok=True)

!cp kaggle.json /root/.kaggle/kaggle.json
!chmod 600 /root/.kaggle/kaggle.json

print("Kaggle authentication successful.")


# ============================================================
# 4. DOWNLOAD EXACT KAGGLE DATASET
# ============================================================

print("\n" + "=" * 70)
print("DOWNLOADING LUNG CANCER DATASET")
print("=" * 70)

!kaggle datasets download \
    -d jillanisofttech/lung-cancer-detection \
    --unzip

print("Dataset downloaded successfully.")


# ============================================================
# 5. FIND CSV FILE
# ============================================================

csv_files = glob.glob(
    "/content/**/*.csv",
    recursive=True
)

if len(csv_files) == 0:
    raise FileNotFoundError(
        "No CSV file was found after downloading the dataset."
    )

print("\nCSV files found:")

for file in csv_files:
    print(file)


# The dataset contains one CSV, so use the first one.
csv_path = csv_files[0]


# ============================================================
# 6. LOAD DATASET USING PANDAS
# ============================================================

df = pd.read_csv(csv_path)

print("\n" + "=" * 70)
print("INITIAL DATASET")
print("=" * 70)

print("Dataset shape:", df.shape)

display(df.head())


# ============================================================
# 7. CLEAN COLUMN NAMES
# ============================================================
# Some columns contain trailing spaces:
#
# FATIGUE
# ALLERGY
#
# Strip whitespace from every column name.
# ============================================================

df.columns = df.columns.str.strip()

print("\nCleaned column names:")
print(df.columns.tolist())


# ============================================================
# 8. CHECK DUPLICATES
# ============================================================

duplicate_count = df.duplicated().sum()

print("\nNumber of duplicate rows:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)

print("Shape after duplicate removal:", df.shape)


# ============================================================
# 9. CHECK MISSING VALUES
# ============================================================

print("\nMissing values:")
print(df.isnull().sum())


# ============================================================
# 10. DEFINE TARGET
# ============================================================

TARGET = "LUNG_CANCER"

if TARGET not in df.columns:
    raise ValueError(
        f"Target column '{TARGET}' was not found."
    )


# ============================================================
# 11. TARGET ENCODING
# ============================================================
# YES -> 1
# NO  -> 0
# ============================================================

df[TARGET] = (
    df[TARGET]
    .astype(str)
    .str.strip()
    .str.upper()
)

target_mapping = {
    "YES": 1,
    "NO": 0
}

df[TARGET] = df[TARGET].map(target_mapping)


# Check for unexpected target values
if df[TARGET].isnull().any():

    raise ValueError(
        "Unexpected values found in LUNG_CANCER column."
    )

df[TARGET] = df[TARGET].astype(int)


# ============================================================
# 12. FEATURE ENGINEERING
# ============================================================
# The dataset already contains engineered risk-factor variables.
#
# We will:
#
# 1. Keep AGE as numerical.
# 2. Keep the symptom/risk variables.
# 3. Convert GENDER through OneHotEncoding.
#
# We also create an additional feature:
#
# RISK_FACTOR_COUNT
#
# It counts how many risk/symptom indicators are active.
#
# This gives the models one aggregated representation of the
# patient's overall symptom/risk burden.
# ============================================================

risk_columns = [
    "SMOKING",
    "YELLOW_FINGERS",
    "ANXIETY",
    "PEER_PRESSURE",
    "CHRONIC DISEASE",
    "FATIGUE",
    "ALLERGY",
    "WHEEZING",
    "ALCOHOL CONSUMING",
    "COUGHING",
    "SHORTNESS OF BREATH",
    "SWALLOWING DIFFICULTY",
    "CHEST PAIN"
]

# Make sure all expected columns exist
missing_risk_columns = [
    col for col in risk_columns
    if col not in df.columns
]

if missing_risk_columns:
    raise ValueError(
        f"Missing expected feature columns: {missing_risk_columns}"
    )


# The dataset uses:
# 1 = No
# 2 = Yes
#
# Convert each feature into:
# 0 = No
# 1 = Yes
#
# This makes the feature representation more intuitive.

for col in risk_columns:

    df[col] = df[col].map({
        1: 0,
        2: 1
    })


# Check whether any unexpected values appeared
for col in risk_columns:

    if df[col].isnull().any():

        raise ValueError(
            f"Unexpected values found in column: {col}"
        )


# Create aggregated risk/symptom feature
df["RISK_FACTOR_COUNT"] = df[risk_columns].sum(axis=1)


print("\nFeature engineering completed.")

print("\nNew feature:")
print("RISK_FACTOR_COUNT")


# ============================================================
# 13. FINAL DATASET INSPECTION
# ============================================================

print("\n" + "=" * 70)
print("FINAL DATASET")
print("=" * 70)

print("Shape:", df.shape)

display(df.head())

print("\nData types:")
print(df.dtypes)


# ============================================================
# 14. CLASS BALANCE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("CLASS BALANCE")
print("=" * 70)

class_counts = df[TARGET].value_counts().sort_index()

print("\nClass counts:")
print(
    class_counts.rename(
        index={
            0: "NO LUNG CANCER",
            1: "LUNG CANCER"
        }
    )
)

class_percentages = (
    df[TARGET]
    .value_counts(normalize=True)
    .sort_index()
    * 100
)

print("\nClass percentages:")

for class_value, percentage in class_percentages.items():

    label = (
        "NO LUNG CANCER"
        if class_value == 0
        else "LUNG CANCER"
    )

    print(
        f"{label}: {percentage:.2f}%"
    )


# Plot class balance

plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x=TARGET
)

plt.title("Original Class Distribution")

plt.xlabel(
    "Lung Cancer (0 = No, 1 = Yes)"
)

plt.ylabel("Number of Samples")

plt.xticks(
    [0, 1],
    ["No Lung Cancer", "Lung Cancer"]
)

plt.tight_layout()

plt.show()


# ============================================================
# 15. SEPARATE FEATURES AND TARGET
# ============================================================

X = df.drop(
    columns=[TARGET]
)

y = df[TARGET]


# ============================================================
# 16. TRAIN / TEST SPLIT
# ============================================================
# 80% training
# 20% testing
#
# stratify=y preserves approximately the same class ratio
# in both sets.
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)


print("\n" + "=" * 70)
print("TRAIN / TEST SPLIT")
print("=" * 70)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

print("\nTraining distribution BEFORE balancing:")

print(
    y_train.value_counts()
    .sort_index()
)


# ============================================================
# 17. IDENTIFY NUMERICAL / CATEGORICAL FEATURES
# ============================================================

numeric_features = X_train.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()


print("\nNumerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)


# ============================================================
# 18. PREPROCESSING
# ============================================================

numeric_pipeline = Pipeline(
    steps=[

        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        )
    ]
)


categorical_pipeline = Pipeline(
    steps=[

        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[

        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),

        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ]
)


# ============================================================
# 19. DEFINE FIVE MODELS
# ============================================================

models = {

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=RANDOM_STATE
    ),

    "SVC": SVC(
        kernel="rbf",
        probability=True,
        random_state=RANDOM_STATE
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5
    ),

    "ANN": MLPClassifier(
        hidden_layer_sizes=(32, 16),
        activation="relu",
        solver="adam",
        alpha=0.0001,
        batch_size=16,
        learning_rate_init=0.001,
        max_iter=500,
        early_stopping=True,
        validation_fraction=0.15,
        n_iter_no_change=20,
        random_state=RANDOM_STATE
    )
}


# ============================================================
# 20. TRAIN + EVALUATE ALL FIVE MODELS
# ============================================================

results = {}

trained_pipelines = {}

predictions = {}

probabilities = {}

confusion_matrices = {}


print("\n" + "=" * 70)
print("TRAINING FIVE MODELS")
print("=" * 70)


for model_name, model in models.items():

    print(
        f"\nTraining {model_name}..."
    )


    # --------------------------------------------------------
    # IMPORTANT:
    #
    # RandomOverSampler is INSIDE the pipeline.
    #
    # Therefore:
    #
    # TRAIN:
    # preprocessing -> oversampling -> model
    #
    # TEST:
    # preprocessing -> model
    #
    # The test data is NEVER oversampled.
    # --------------------------------------------------------

    pipeline = ImbPipeline(
        steps=[

            (
                "preprocessor",
                preprocessor
            ),

            (
                "balancer",
                RandomOverSampler(
                    random_state=RANDOM_STATE
                )
            ),

            (
                "model",
                model
            )
        ]
    )


    # Train
    pipeline.fit(
        X_train,
        y_train
    )


    # Predict class
    y_pred = pipeline.predict(
        X_test
    )


    # Predict probability
    y_prob = pipeline.predict_proba(
        X_test
    )[:, 1]


    # --------------------------------------------------------
    # METRICS
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_test,
        y_prob
    )


    # Store results
    results[model_name] = {

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "ROC-AUC": roc_auc
    }


    trained_pipelines[
        model_name
    ] = pipeline


    predictions[
        model_name
    ] = y_pred


    probabilities[
        model_name
    ] = y_prob


    confusion_matrices[
        model_name
    ] = confusion_matrix(
        y_test,
        y_pred
    )


    print(
        f"Accuracy  : {accuracy:.4f}"
    )

    print(
        f"Precision : {precision:.4f}"
    )

    print(
        f"Recall    : {recall:.4f}"
    )

    print(
        f"ROC-AUC   : {roc_auc:.4f}"
    )


# ============================================================
# 21. PERFORMANCE TABLE
# ============================================================

results_df = pd.DataFrame(
    results
).T


print("\n" + "=" * 70)
print("MODEL PERFORMANCE COMPARISON")
print("=" * 70)

display(
    results_df.round(4)
)


# ============================================================
# 22. CALCULATE AVERAGE SCORE
# ============================================================
# Equal-weight average of:
#
# Accuracy
# Precision
# Recall
# ROC-AUC
#
# This is used only as the project's model-selection criterion.
# ============================================================

metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "ROC-AUC"
]

results_df[
    "Average Score"
] = results_df[
    metric_columns
].mean(
    axis=1
)


print("\n" + "=" * 70)
print("AVERAGE SCORE")
print("=" * 70)

display(
    results_df
    .sort_values(
        "Average Score",
        ascending=False
    )
    .round(4)
)


# ============================================================
# 23. CONFUSION MATRIX HEATMAPS
# ============================================================

print("\n" + "=" * 70)
print("CONFUSION MATRICES")
print("=" * 70)


for model_name, cm in confusion_matrices.items():

    plt.figure(
        figsize=(5, 4)
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        cbar=False,
        xticklabels=[
            "No Lung Cancer",
            "Lung Cancer"
        ],
        yticklabels=[
            "No Lung Cancer",
            "Lung Cancer"
        ]
    )

    plt.title(
        f"Confusion Matrix - {model_name}"
    )

    plt.xlabel(
        "Predicted"
    )

    plt.ylabel(
        "Actual"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 24. COMPARATIVE GRAPH - ALL FOUR METRICS
# ============================================================

plt.figure(
    figsize=(13, 6)
)

results_df[
    metric_columns
].plot(
    kind="bar",
    figsize=(13, 6)
)

plt.title(
    "Comparison of Five Models Across All Evaluation Metrics"
)

plt.xlabel(
    "Machine Learning Model"
)

plt.ylabel(
    "Score"
)

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=0
)

plt.legend(
    title="Metrics"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================
# 25. INDIVIDUAL METRIC COMPARISON
# ============================================================

for metric in metric_columns:

    plt.figure(
        figsize=(10, 5)
    )

    values = results_df[
        metric
    ].sort_values(
        ascending=False
    )

    bars = plt.bar(
        values.index,
        values.values
    )

    plt.title(
        f"{metric} Comparison"
    )

    plt.xlabel(
        "Model"
    )

    plt.ylabel(
        metric
    )

    plt.ylim(
        0,
        1.05
    )

    plt.xticks(
        rotation=0
    )

    plt.grid(
        axis="y",
        alpha=0.3
    )


    # Add values above bars

    for bar, value in zip(
        bars,
        values.values
    ):

        plt.text(
            bar.get_x()
            + bar.get_width() / 2,

            value + 0.02,

            f"{value:.3f}",

            ha="center"
        )


    plt.tight_layout()

    plt.show()


# ============================================================
# 26. AVERAGE SCORE GRAPH
# ============================================================

average_scores = results_df[
    "Average Score"
].sort_values(
    ascending=False
)


plt.figure(
    figsize=(10, 5)
)

bars = plt.bar(
    average_scores.index,
    average_scores.values
)

plt.title(
    "Average Performance Across Accuracy, Precision, Recall and ROC-AUC"
)

plt.xlabel(
    "Model"
)

plt.ylabel(
    "Average Score"
)

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=0
)

plt.grid(
    axis="y",
    alpha=0.3
)


for bar, value in zip(
    bars,
    average_scores.values
):

    plt.text(
        bar.get_x()
        + bar.get_width() / 2,

        value + 0.02,

        f"{value:.3f}",

        ha="center"
    )


plt.tight_layout()

plt.show()


# ============================================================
# 27. SELECT BEST MODEL
# ============================================================

best_model_name = results_df[
    "Average Score"
].idxmax()


best_model_pipeline = trained_pipelines[
    best_model_name
]


best_score = results_df.loc[
    best_model_name,
    "Average Score"
]


print("\n" + "=" * 70)
print("BEST MODEL")
print("=" * 70)

print(
    "Selected model:",
    best_model_name
)

print(
    f"Average score: {best_score:.4f}"
)


# ============================================================
# 28. DISPLAY BEST MODEL'S METRICS
# ============================================================

print("\nBest model metrics:")

for metric in metric_columns:

    print(
        f"{metric:10s}: "
        f"{results_df.loc[best_model_name, metric]:.4f}"
    )


# ============================================================
# 29. SAVE BEST MODEL USING PICKLE
# ============================================================

MODEL_FILENAME = (
    "lung_cancer_best_model.pkl"
)


with open(
    MODEL_FILENAME,
    "wb"
) as file:

    pickle.dump(
        best_model_pipeline,
        file
    )


print("\nBest model saved successfully:")
print(
    MODEL_FILENAME
)


# ============================================================
# 30. SAVE COMPLETE RESULTS
# ============================================================

RESULTS_FILENAME = (
    "lung_cancer_model_comparison.csv"
)


results_df.sort_values(
    "Average Score",
    ascending=False
).to_csv(
    RESULTS_FILENAME
)


print(
    "Model comparison saved:"
)

print(
    RESULTS_FILENAME
)


# ============================================================
# 31. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL PROJECT SUMMARY")
print("=" * 70)

print(
    f"Original dataset size : 309 rows"
)

print(
    f"Final dataset size    : {len(df)} rows"
)

print(
    f"Number of features    : {X.shape[1]}"
)

print(
    f"Training samples      : {len(X_train)}"
)

print(
    f"Testing samples       : {len(X_test)}"
)

print(
    f"Best model            : {best_model_name}"
)

print(
    f"Average score         : {best_score:.4f}"
)

print("\nBest model metrics:")

for metric in metric_columns:

    print(
        f"{metric:10s}: "
        f"{results_df.loc[best_model_name, metric]:.4f}"
    )

print("\nGenerated files:")

print(
    "-",
    MODEL_FILENAME
)

print(
    "-",
    RESULTS_FILENAME
)

print("\nLUNG CANCER PIPELINE COMPLETED.")
```

```python
# ============================================================
# LIVER DISEASE PREDICTION - COMPLETE ML PIPELINE
# ============================================================
# Dataset:
# rabieelkharoua/predict-liver-disease-1700-records-dataset
#
# Models:
#   1. Random Forest
#   2. Logistic Regression
#   3. SVC
#   4. KNN
#   5. ANN
#
# Evaluation:
#   Accuracy
#   Precision
#   Recall
#   ROC-AUC
#
# Additional:
#   - Data cleaning
#   - Feature engineering
#   - Class balance analysis
#   - Training-only oversampling
#   - Confusion matrices
#   - Comparative graphs
#   - Best-model selection
#   - Pickle
# ============================================================


# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

!pip -q install kaggle imbalanced-learn


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import os
import glob
import pickle
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestClassifier

from sklearn.linear_model import LogisticRegression

from sklearn.svm import SVC

from sklearn.neighbors import KNeighborsClassifier

from sklearn.neural_network import MLPClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    roc_auc_score,
    confusion_matrix
)

from imblearn.over_sampling import RandomOverSampler
from imblearn.pipeline import Pipeline as ImbPipeline

warnings.filterwarnings("ignore")

RANDOM_STATE = 42


# ============================================================
# 3. KAGGLE AUTHENTICATION
# ============================================================

print("=" * 70)
print("KAGGLE AUTHENTICATION")
print("=" * 70)

uploaded = files.upload()

os.makedirs("/root/.kaggle", exist_ok=True)

!cp kaggle.json /root/.kaggle/kaggle.json
!chmod 600 /root/.kaggle/kaggle.json

print("Kaggle authentication successful.")


# ============================================================
# 4. DOWNLOAD EXACT DATASET
# ============================================================

print("\n" + "=" * 70)
print("DOWNLOADING LIVER DISEASE DATASET")
print("=" * 70)

!kaggle datasets download \
    -d rabieelkharoua/predict-liver-disease-1700-records-dataset \
    --unzip

print("Dataset downloaded successfully.")


# ============================================================
# 5. FIND CSV FILE
# ============================================================

csv_files = glob.glob(
    "/content/**/*.csv",
    recursive=True
)

if len(csv_files) == 0:
    raise FileNotFoundError(
        "No CSV file was found after downloading the dataset."
    )

print("\nCSV files found:")

for file in csv_files:
    print(file)


# ============================================================
# 6. LOAD DATASET
# ============================================================

csv_path = csv_files[0]

df = pd.read_csv(csv_path)

print("\n" + "=" * 70)
print("INITIAL DATASET")
print("=" * 70)

print("Dataset shape:", df.shape)

display(df.head())


# ============================================================
# 7. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("/", "_")
)

print("\nCleaned column names:")
print(df.columns.tolist())


# ============================================================
# 8. CHECK DUPLICATES
# ============================================================

duplicate_count = df.duplicated().sum()

print(
    "\nDuplicate rows:",
    duplicate_count
)

if duplicate_count > 0:

    df = (
        df
        .drop_duplicates()
        .reset_index(drop=True)
    )

print(
    "Shape after duplicate removal:",
    df.shape
)


# ============================================================
# 9. CHECK MISSING VALUES
# ============================================================

print("\nMissing values:")

print(
    df.isnull().sum()
)


# ============================================================
# 10. IDENTIFY TARGET COLUMN
# ============================================================
# The 1700-record dataset uses a categorical liver-disease
# target representing whether liver disease is present.
#
# Because Kaggle dataset copies can sometimes use slightly
# different capitalization/spacing, we identify the target
# robustly.
# ============================================================

possible_targets = [
    "Liver_Disease",
    "LiverDisease",
    "Diagnosis",
    "Target",
    "Result",
    "Outcome",
    "Dataset",
    "Disease"
]

TARGET = None

for col in possible_targets:

    if col in df.columns:

        TARGET = col
        break


if TARGET is None:

    # Fallback:
    # look for a column containing "liver"

    liver_columns = [
        col
        for col in df.columns
        if "liver" in col.lower()
    ]

    if len(liver_columns) > 0:

        TARGET = liver_columns[-1]


if TARGET is None:

    raise ValueError(
        "Could not automatically identify the liver-disease "
        "target column.\n\nAvailable columns:\n"
        + str(df.columns.tolist())
    )


print(
    "\nTarget column:",
    TARGET
)


# ============================================================
# 11. TARGET INSPECTION
# ============================================================

print("\nTarget values:")

print(
    df[TARGET].value_counts(
        dropna=False
    )
)


# ============================================================
# 12. TARGET ENCODING
# ============================================================
# Convert common representations into:
#
# 0 = No liver disease
# 1 = Liver disease
#
# The code handles several possible Kaggle encodings.
# ============================================================

if df[TARGET].dtype == "object":

    df[TARGET] = (
        df[TARGET]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    unique_values = set(
        df[TARGET].dropna().unique()
    )

    print(
        "\nUnique target values:",
        unique_values
    )

    target_mapping = {

        # Positive class
        "yes": 1,
        "y": 1,
        "present": 1,
        "positive": 1,
        "liver disease": 1,
        "liver_disease": 1,
        "disease": 1,
        "1": 1,

        # Negative class
        "no": 0,
        "n": 0,
        "absent": 0,
        "negative": 0,
        "no liver disease": 0,
        "no_liver_disease": 0,
        "healthy": 0,
        "0": 0
    }

    df[TARGET] = df[TARGET].map(
        target_mapping
    )

else:

    # Numeric targets are converted directly
    df[TARGET] = pd.to_numeric(
        df[TARGET],
        errors="coerce"
    )


# ------------------------------------------------------------
# Handle datasets where disease labels are 1/2
#
# If exactly two numeric classes exist and they are 1 and 2:
#   1 -> 0
#   2 -> 1
# ------------------------------------------------------------

unique_target_values = sorted(
    df[TARGET]
    .dropna()
    .unique()
)


if set(unique_target_values) == {1, 2}:

    df[TARGET] = df[TARGET].map({
        1: 0,
        2: 1
    })


# Remove rows with unknown target values
df = df.dropna(
    subset=[TARGET]
).reset_index(drop=True)

df[TARGET] = df[TARGET].astype(int)


# ============================================================
# 13. VERIFY BINARY TARGET
# ============================================================

unique_target_values = sorted(
    df[TARGET].unique()
)

if not set(unique_target_values).issubset({0, 1}):

    raise ValueError(
        "Target could not be converted into a binary "
        "0/1 classification problem.\n"
        f"Detected values: {unique_target_values}"
    )


print("\nFinal target distribution:")

print(
    df[TARGET].value_counts()
    .sort_index()
)


# ============================================================
# 14. FEATURE ENGINEERING
# ============================================================
#
# This dataset primarily consists of clinical biochemical
# measurements:
#
#   Age
#   Gender
#   Total Bilirubin
#   Direct Bilirubin
#   Alkaline Phosphatase
#   ALT
#   AST
#   Total Proteins
#   Albumin
#   Albumin/Globulin Ratio
#
# We create clinically meaningful derived ratios while keeping
# the original measurements.
#
# IMPORTANT:
# These are mathematical feature transformations, NOT medical
# diagnostic rules.
# ============================================================


# ------------------------------------------------------------
# Identify columns automatically
# ------------------------------------------------------------

def find_column(possible_names):

    normalized = {
        col.lower()
        .replace("_", "")
        .replace(" ", "")
        .replace("-", ""): col
        for col in df.columns
    }

    for name in possible_names:

        key = (
            name.lower()
            .replace("_", "")
            .replace(" ", "")
            .replace("-", "")
        )

        if key in normalized:

            return normalized[key]

    return None


total_bilirubin_col = find_column([
    "Total_Bilirubin",
    "Total Bilirubin",
    "TotalBilirubin"
])

direct_bilirubin_col = find_column([
    "Direct_Bilirubin",
    "Direct Bilirubin",
    "DirectBilirubin"
])

albumin_col = find_column([
    "Albumin"
])

ratio_col = find_column([
    "Albumin_and_Globulin_Ratio",
    "Albumin_Globulin_Ratio",
    "A_G_Ratio",
    "AG_Ratio"
])

alt_col = find_column([
    "ALT",
    "Alamine_Aminotransferase",
    "Alamine_Aminotransferase"
])

ast_col = find_column([
    "AST",
    "Aspartate_Aminotransferase"
])


# ------------------------------------------------------------
# Derived feature 1:
#
# Direct / Total Bilirubin ratio
# ------------------------------------------------------------

if (
    total_bilirubin_col is not None
    and direct_bilirubin_col is not None
):

    df[
        "Direct_Total_Bilirubin_Ratio"
    ] = (
        df[direct_bilirubin_col]
        /
        df[total_bilirubin_col].replace(
            0,
            np.nan
        )
    )


# ------------------------------------------------------------
# Derived feature 2:
#
# ALT / AST ratio
# ------------------------------------------------------------

if (
    alt_col is not None
    and ast_col is not None
):

    df[
        "ALT_AST_Ratio"
    ] = (
        df[alt_col]
        /
        df[ast_col].replace(
            0,
            np.nan
        )
    )


# ------------------------------------------------------------
# Derived feature 3:
#
# Albumin relative to total proteins
# ------------------------------------------------------------

if (
    albumin_col is not None
):

    total_protein_col = find_column([
        "Total_Proteins",
        "Total Protiens",
        "Total Proteins",
        "TotalProtein"
    ])

    if total_protein_col is not None:

        df[
            "Albumin_Protein_Ratio"
        ] = (
            df[albumin_col]
            /
            df[total_protein_col].replace(
                0,
                np.nan
            )
        )


print("\nFeature engineering completed.")

print("\nNew/derived features:")

derived_features = [
    col
    for col in df.columns
    if col in [
        "Direct_Total_Bilirubin_Ratio",
        "ALT_AST_Ratio",
        "Albumin_Protein_Ratio"
    ]
]

print(derived_features)


# ============================================================
# 15. CLEAN INFINITE VALUES
# ============================================================

df = df.replace(
    [np.inf, -np.inf],
    np.nan
)


# ============================================================
# 16. CLASS BALANCE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("CLASS BALANCE")
print("=" * 70)

class_counts = (
    df[TARGET]
    .value_counts()
    .sort_index()
)

print("\nClass counts:")

for class_value, count in class_counts.items():

    label = (
        "No Liver Disease"
        if class_value == 0
        else "Liver Disease"
    )

    print(
        f"{label}: {count}"
    )


class_percentages = (
    df[TARGET]
    .value_counts(
        normalize=True
    )
    .sort_index()
    * 100
)


print("\nClass percentages:")

for class_value, percentage in class_percentages.items():

    label = (
        "No Liver Disease"
        if class_value == 0
        else "Liver Disease"
    )

    print(
        f"{label}: {percentage:.2f}%"
    )


# ============================================================
# 17. CLASS DISTRIBUTION GRAPH
# ============================================================

plt.figure(
    figsize=(6, 4)
)

sns.countplot(
    data=df,
    x=TARGET
)

plt.title(
    "Original Liver Disease Class Distribution"
)

plt.xlabel(
    "Liver Disease"
)

plt.ylabel(
    "Number of Patients"
)

plt.xticks(
    [0, 1],
    [
        "No Liver Disease",
        "Liver Disease"
    ]
)

plt.tight_layout()

plt.show()


# ============================================================
# 18. SEPARATE FEATURES AND TARGET
# ============================================================

X = df.drop(
    columns=[TARGET]
)

y = df[TARGET]


# ============================================================
# 19. TRAIN / TEST SPLIT
# ============================================================
#
# IMPORTANT:
# The split occurs BEFORE oversampling.
#
# This prevents synthetic/duplicated training samples from
# leaking into the test set.
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=RANDOM_STATE,

    stratify=y
)


print("\n" + "=" * 70)
print("TRAIN / TEST SPLIT")
print("=" * 70)

print(
    "Training samples:",
    len(X_train)
)

print(
    "Testing samples :",
    len(X_test)
)

print(
    "\nTraining class distribution BEFORE balancing:"
)

print(
    y_train.value_counts()
    .sort_index()
)


# ============================================================
# 20. IDENTIFY FEATURE TYPES
# ============================================================

numeric_features = (
    X_train
    .select_dtypes(
        include=[
            "int64",
            "float64",
            "int32",
            "float32"
        ]
    )
    .columns
    .tolist()
)


categorical_features = (
    X_train
    .select_dtypes(
        include=[
            "object",
            "category",
            "bool"
        ]
    )
    .columns
    .tolist()
)


print("\nNumerical features:")

print(
    numeric_features
)


print("\nCategorical features:")

print(
    categorical_features
)


# ============================================================
# 21. PREPROCESSING
# ============================================================

numeric_pipeline = Pipeline(
    steps=[

        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        )
    ]
)


categorical_pipeline = Pipeline(
    steps=[

        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[

        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),

        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ]
)


# ============================================================
# 22. DEFINE FIVE MODELS
# ============================================================

models = {

    "Random Forest":

        RandomForestClassifier(

            n_estimators=300,

            random_state=RANDOM_STATE,

            n_jobs=-1
        ),


    "Logistic Regression":

        LogisticRegression(

            max_iter=2000,

            random_state=RANDOM_STATE
        ),


    "SVC":

        SVC(

            kernel="rbf",

            probability=True,

            random_state=RANDOM_STATE
        ),


    "KNN":

        KNeighborsClassifier(

            n_neighbors=7
        ),


    "ANN":

        MLPClassifier(

            hidden_layer_sizes=(
                64,
                32
            ),

            activation="relu",

            solver="adam",

            alpha=0.0001,

            batch_size=32,

            learning_rate_init=0.001,

            max_iter=500,

            early_stopping=True,

            validation_fraction=0.15,

            n_iter_no_change=20,

            random_state=RANDOM_STATE
        )
}


# ============================================================
# 23. TRAIN + EVALUATE ALL MODELS
# ============================================================

results = {}

trained_pipelines = {}

predictions = {}

probabilities = {}

confusion_matrices = {}


print("\n" + "=" * 70)
print("MODEL TRAINING")
print("=" * 70)


for model_name, model in models.items():

    print(
        f"\nTraining {model_name}..."
    )


    # --------------------------------------------------------
    # Oversampling is inside the pipeline.
    #
    # It therefore happens ONLY to training data.
    # --------------------------------------------------------

    pipeline = ImbPipeline(
        steps=[

            (
                "preprocessor",
                preprocessor
            ),

            (
                "balancer",
                RandomOverSampler(
                    random_state=RANDOM_STATE
                )
            ),

            (
                "model",
                model
            )
        ]
    )


    # Train
    pipeline.fit(
        X_train,
        y_train
    )


    # Predict
    y_pred = pipeline.predict(
        X_test
    )


    # Probability
    y_prob = pipeline.predict_proba(
        X_test
    )[:, 1]


    # --------------------------------------------------------
    # METRICS
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_test,
        y_pred
    )


    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )


    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )


    roc_auc = roc_auc_score(
        y_test,
        y_prob
    )


    # --------------------------------------------------------
    # STORE RESULTS
    # --------------------------------------------------------

    results[model_name] = {

        "Accuracy":
            accuracy,

        "Precision":
            precision,

        "Recall":
            recall,

        "ROC-AUC":
            roc_auc
    }


    trained_pipelines[
        model_name
    ] = pipeline


    predictions[
        model_name
    ] = y_pred


    probabilities[
        model_name
    ] = y_prob


    confusion_matrices[
        model_name
    ] = confusion_matrix(
        y_test,
        y_pred
    )


    print(
        f"Accuracy  : {accuracy:.4f}"
    )

    print(
        f"Precision : {precision:.4f}"
    )

    print(
        f"Recall    : {recall:.4f}"
    )

    print(
        f"ROC-AUC   : {roc_auc:.4f}"
    )


# ============================================================
# 24. RESULTS TABLE
# ============================================================

results_df = pd.DataFrame(
    results
).T


print("\n" + "=" * 70)
print("MODEL PERFORMANCE")
print("=" * 70)

display(
    results_df.round(4)
)


# ============================================================
# 25. AVERAGE SCORE
# ============================================================

metric_columns = [

    "Accuracy",

    "Precision",

    "Recall",

    "ROC-AUC"
]


results_df[
    "Average Score"
] = results_df[
    metric_columns
].mean(
    axis=1
)


print("\n" + "=" * 70)
print("AVERAGE MODEL SCORE")
print("=" * 70)

display(
    results_df
    .sort_values(
        "Average Score",
        ascending=False
    )
    .round(4)
)


# ============================================================
# 26. CONFUSION MATRIX HEATMAPS
# ============================================================

print("\n" + "=" * 70)
print("CONFUSION MATRICES")
print("=" * 70)


for model_name, cm in confusion_matrices.items():

    plt.figure(
        figsize=(5, 4)
    )


    sns.heatmap(

        cm,

        annot=True,

        fmt="d",

        cmap="Blues",

        cbar=False,

        xticklabels=[
            "No Disease",
            "Disease"
        ],

        yticklabels=[
            "No Disease",
            "Disease"
        ]
    )


    plt.title(
        f"Confusion Matrix - {model_name}"
    )


    plt.xlabel(
        "Predicted"
    )


    plt.ylabel(
        "Actual"
    )


    plt.tight_layout()

    plt.show()


# ============================================================
# 27. COMPARATIVE GRAPH - FOUR METRICS
# ============================================================

plt.figure(
    figsize=(13, 6)
)

results_df[
    metric_columns
].plot(
    kind="bar",
    figsize=(13, 6)
)

plt.title(
    "Comparison of Five Models Across Evaluation Metrics"
)

plt.xlabel(
    "Model"
)

plt.ylabel(
    "Score"
)

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=0
)

plt.legend(
    title="Metrics"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================
# 28. INDIVIDUAL METRIC GRAPHS
# ============================================================

for metric in metric_columns:

    values = (
        results_df[metric]
        .sort_values(
            ascending=False
        )
    )


    plt.figure(
        figsize=(10, 5)
    )


    bars = plt.bar(
        values.index,
        values.values
    )


    plt.title(
        f"{metric} Comparison"
    )


    plt.xlabel(
        "Model"
    )


    plt.ylabel(
        metric
    )


    plt.ylim(
        0,
        1.05
    )


    plt.xticks(
        rotation=0
    )


    plt.grid(
        axis="y",
        alpha=0.3
    )


    for bar, value in zip(
        bars,
        values.values
    ):

        plt.text(

            bar.get_x()
            + bar.get_width() / 2,

            value + 0.02,

            f"{value:.3f}",

            ha="center"
        )


    plt.tight_layout()

    plt.show()


# ============================================================
# 29. AVERAGE SCORE GRAPH
# ============================================================

average_scores = (
    results_df[
        "Average Score"
    ]
    .sort_values(
        ascending=False
    )
)


plt.figure(
    figsize=(10, 5)
)


bars = plt.bar(

    average_scores.index,

    average_scores.values
)


plt.title(
    "Average Performance Across All Four Metrics"
)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "Average Score"
)


plt.ylim(
    0,
    1.05
)


plt.xticks(
    rotation=0
)


plt.grid(
    axis="y",
    alpha=0.3
)


for bar, value in zip(
    bars,
    average_scores.values
):

    plt.text(

        bar.get_x()
        + bar.get_width() / 2,

        value + 0.02,

        f"{value:.3f}",

        ha="center"
    )


plt.tight_layout()

plt.show()


# ============================================================
# 30. SELECT BEST MODEL
# ============================================================

best_model_name = (
    results_df[
        "Average Score"
    ]
    .idxmax()
)


best_model_pipeline = (
    trained_pipelines[
        best_model_name
    ]
)


best_score = (
    results_df.loc[
        best_model_name,
        "Average Score"
    ]
)


print("\n" + "=" * 70)
print("BEST MODEL")
print("=" * 70)


print(
    "Selected model:",
    best_model_name
)


print(
    f"Average score: {best_score:.4f}"
)


# ============================================================
# 31. DISPLAY BEST MODEL METRICS
# ============================================================

print("\nBest model metrics:")


for metric in metric_columns:

    print(

        f"{metric:10s}: "

        f"{results_df.loc[best_model_name, metric]:.4f}"

    )


# ============================================================
# 32. SAVE BEST MODEL USING PICKLE
# ============================================================

MODEL_FILENAME = (
    "liver_disease_best_model.pkl"
)


with open(
    MODEL_FILENAME,
    "wb"
) as file:

    pickle.dump(
        best_model_pipeline,
        file
    )


print(
    "\nBest model saved successfully:"
)

print(
    MODEL_FILENAME
)


# ============================================================
# 33. SAVE MODEL COMPARISON
# ============================================================

RESULTS_FILENAME = (
    "liver_disease_model_comparison.csv"
)


results_df.sort_values(
    "Average Score",
    ascending=False
).to_csv(
    RESULTS_FILENAME
)


print(
    "\nComparison results saved:"
)

print(
    RESULTS_FILENAME
)


# ============================================================
# 34. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL LIVER DISEASE PROJECT SUMMARY")
print("=" * 70)


print(
    f"Final dataset size : {len(df)} rows"
)


print(
    f"Number of features: {X.shape[1]}"
)


print(
    f"Training samples   : {len(X_train)}"
)


print(
    f"Testing samples    : {len(X_test)}"
)


print(
    f"Best model         : {best_model_name}"
)


print(
    f"Average score      : {best_score:.4f}"
)


print("\nBest model metrics:")


for metric in metric_columns:

    print(

        f"{metric:10s}: "

        f"{results_df.loc[best_model_name, metric]:.4f}"

    )


print("\nGenerated files:")

print(
    "-",
    MODEL_FILENAME
)

print(
    "-",
    RESULTS_FILENAME
)


print(
    "\nLIVER DISEASE PIPELINE COMPLETED."
)
```

```python
# ============================================================
# DATASET 3: STROKE PREDICTION
# Kaggle: samueltaiwograce/stroke-dataset
# ============================================================

# -----------------------------
# 1. Install dependencies
# -----------------------------
!pip -q install kaggle imbalanced-learn

# -----------------------------
# 2. Imports
# -----------------------------
import os
import glob
import pickle
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    roc_auc_score,
    confusion_matrix
)

from imblearn.over_sampling import RandomOverSampler
from imblearn.pipeline import Pipeline as ImbPipeline

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

# ============================================================
# 3. Kaggle authentication
# ============================================================
print("Upload your kaggle.json file:")

uploaded = files.upload()

os.makedirs("/root/.kaggle", exist_ok=True)

# Colab may rename repeated uploads:
# kaggle.json -> kaggle (1).json -> kaggle (2).json -> ...
kaggle_file = None

for filename in uploaded.keys():
    if filename.lower().endswith(".json"):
        kaggle_file = filename
        break

if kaggle_file is None:
    raise FileNotFoundError(
        "No .json file was uploaded. Please upload kaggle.json."
    )

with open("/root/.kaggle/kaggle.json", "wb") as f:
    f.write(uploaded[kaggle_file])

os.chmod("/root/.kaggle/kaggle.json", 0o600)

print(f"Kaggle authentication configured using: {kaggle_file}")

# ============================================================
# 4. Download EXACT stroke dataset
# ============================================================

DOWNLOAD_DIR = "/content/stroke_dataset"

os.makedirs(DOWNLOAD_DIR, exist_ok=True)

print("\nDownloading stroke dataset...")

!kaggle datasets download \
    -d samueltaiwograce/stroke-dataset \
    -p /content/stroke_dataset \
    --unzip

print("Download complete.")

# ============================================================
# 5. Locate CSV robustly
# ============================================================

csv_files = glob.glob(
    os.path.join(DOWNLOAD_DIR, "**", "*.csv"),
    recursive=True
)

if len(csv_files) == 0:
    raise FileNotFoundError(
        "No CSV file found inside the downloaded Kaggle dataset."
    )

print("\nCSV files found:")
for f in csv_files:
    print(" -", f)

# Try to identify the actual stroke CSV using expected column names.
selected_csv = None

for f in csv_files:
    try:
        temp = pd.read_csv(f, nrows=5)
        cols = {
            str(c).strip().lower().replace(" ", "_")
            for c in temp.columns
        }

        # Expected stroke-dataset indicators
        indicators = {
            "stroke",
            "age",
            "gender",
            "hypertension",
            "heart_disease",
            "avg_glucose_level",
            "bmi"
        }

        if len(indicators.intersection(cols)) >= 4:
            selected_csv = f
            break

    except Exception:
        pass

if selected_csv is None:
    selected_csv = csv_files[0]

print("\nUsing CSV:")
print(selected_csv)

df = pd.read_csv(selected_csv)

print("\nOriginal shape:", df.shape)

# ============================================================
# 6. Basic cleaning
# ============================================================

# Clean column names
df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(" ", "_", regex=False)
    .str.replace("/", "_", regex=False)
)

print("\nColumns:")
print(df.columns.tolist())

# Remove completely empty columns
df = df.dropna(axis=1, how="all")

# Remove duplicate rows
before_duplicates = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after_duplicates = len(df)

print(
    f"\nDuplicate rows removed: "
    f"{before_duplicates - after_duplicates}"
)

print("Shape after cleaning:", df.shape)

# ============================================================
# 7. Detect target column
# ============================================================

possible_targets = [
    "stroke",
    "Stroke",
    "STROKE",
    "target",
    "Target",
    "outcome",
    "Outcome",
    "diagnosis",
    "Diagnosis"
]

target_col = None

for col in possible_targets:
    if col in df.columns:
        target_col = col
        break

if target_col is None:

    # fallback: search case-insensitively
    for col in df.columns:
        if str(col).strip().lower() == "stroke":
            target_col = col
            break

if target_col is None:
    raise ValueError(
        "Could not automatically identify the stroke target column.\n"
        f"Available columns: {df.columns.tolist()}"
    )

print("\nTarget column:", target_col)

# ============================================================
# 8. Clean target
# ============================================================

print("\nOriginal target values:")
print(df[target_col].value_counts(dropna=False))

# Convert target robustly
if df[target_col].dtype == "object":

    target_string = (
        df[target_col]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    target_map = {
        "yes": 1,
        "y": 1,
        "true": 1,
        "positive": 1,
        "stroke": 1,
        "1": 1,

        "no": 0,
        "n": 0,
        "false": 0,
        "negative": 0,
        "no stroke": 0,
        "0": 0
    }

    df[target_col] = target_string.map(target_map)

else:
    df[target_col] = pd.to_numeric(
        df[target_col],
        errors="coerce"
    )

# Remove rows where target couldn't be interpreted
df = df.dropna(subset=[target_col]).copy()

df[target_col] = df[target_col].astype(int)

# Handle possible 1/2 encoding
unique_target = sorted(df[target_col].unique().tolist())

if unique_target == [1, 2]:
    df[target_col] = df[target_col].map({
        1: 0,
        2: 1
    })

# Final binary validation
unique_target = sorted(df[target_col].unique().tolist())

if unique_target != [0, 1]:
    raise ValueError(
        f"Target is not binary 0/1 after cleaning. "
        f"Found: {unique_target}"
    )

print("\nCleaned target distribution:")
print(df[target_col].value_counts())

# ============================================================
# 9. Convert common categorical values
# ============================================================

# Standardize categorical string values
for col in df.select_dtypes(include=["object"]).columns:
    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
        .replace({
            "nan": np.nan,
            "None": np.nan,
            "none": np.nan
        })
    )

# ============================================================
# 10. Feature engineering
# ============================================================

print("\nPerforming feature engineering...")

# Age groups
if "age" in df.columns:
    df["age_squared"] = df["age"] ** 2

    df["age_group"] = pd.cut(
        df["age"],
        bins=[-np.inf, 18, 40, 60, np.inf],
        labels=[
            "Young",
            "Adult",
            "Middle_Aged",
            "Senior"
        ]
    )

# Glucose-related feature
if "avg_glucose_level" in df.columns:
    glucose_median = df["avg_glucose_level"].median()

    df["high_glucose_flag"] = (
        df["avg_glucose_level"] > glucose_median
    ).astype(int)

# BMI-related feature
if "bmi" in df.columns:

    bmi_median = df["bmi"].median()

    df["high_bmi_flag"] = (
        df["bmi"] > bmi_median
    ).astype(int)

# Cardiovascular risk count
risk_cols = [
    "hypertension",
    "heart_disease"
]

existing_risk_cols = [
    col for col in risk_cols
    if col in df.columns
]

if existing_risk_cols:
    df["cardiovascular_risk_count"] = (
        df[existing_risk_cols]
        .fillna(0)
        .sum(axis=1)
    )

# Age × hypertension interaction
if (
    "age" in df.columns
    and "hypertension" in df.columns
):
    df["age_hypertension_interaction"] = (
        df["age"] *
        df["hypertension"].fillna(0)
    )

# Age × heart disease interaction
if (
    "age" in df.columns
    and "heart_disease" in df.columns
):
    df["age_heart_disease_interaction"] = (
        df["age"] *
        df["heart_disease"].fillna(0)
    )

# Glucose × BMI interaction
if (
    "avg_glucose_level" in df.columns
    and "bmi" in df.columns
):
    df["glucose_bmi_interaction"] = (
        df["avg_glucose_level"] *
        df["bmi"]
    )

print("Feature engineering complete.")
print("New shape:", df.shape)

# ============================================================
# 11. Missing-value inspection
# ============================================================

missing = df.isnull().sum()

missing = missing[missing > 0].sort_values(
    ascending=False
)

print("\nMissing values:")
if len(missing) == 0:
    print("No missing values.")
else:
    print(missing)

# ============================================================
# 12. Class balance BEFORE splitting
# ============================================================

class_counts = df[target_col].value_counts().sort_index()

print("\nClass distribution:")
print(class_counts)

class_percent = (
    class_counts /
    class_counts.sum() *
    100
)

print("\nClass percentages:")
for cls, pct in class_percent.items():
    label = "No Stroke" if cls == 0 else "Stroke"
    print(f"{label}: {pct:.2f}%")

plt.figure(figsize=(7, 5))

sns.countplot(
    x=target_col,
    data=df
)

plt.title("Stroke Dataset - Original Class Distribution")
plt.xlabel("Stroke (0 = No Stroke, 1 = Stroke)")
plt.ylabel("Number of Samples")

plt.show()

# ============================================================
# 13. Separate features and target
# ============================================================

X = df.drop(columns=[target_col])
y = df[target_col]

# Remove ID column if present
id_like_columns = [
    col for col in X.columns
    if str(col).strip().lower() in [
        "id",
        "patient_id",
        "patientid"
    ]
]

if id_like_columns:
    print("\nRemoving identifier columns:", id_like_columns)
    X = X.drop(columns=id_like_columns)

# ============================================================
# 14. TRAIN / TEST SPLIT
# IMPORTANT:
# Split BEFORE balancing.
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("\nTrain shape:", X_train.shape)
print("Test shape:", X_test.shape)

print("\nTraining class distribution BEFORE balancing:")
print(y_train.value_counts())

print("\nTest class distribution:")
print(y_test.value_counts())

# ============================================================
# 15. Determine numeric / categorical columns
# ============================================================

numeric_features = X_train.select_dtypes(
    include=["int64", "int32", "float64", "float32"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nNumeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)

# ============================================================
# 16. Preprocessing
# ============================================================

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

transformers = []

if numeric_features:
    transformers.append(
        (
            "num",
            numeric_pipeline,
            numeric_features
        )
    )

if categorical_features:
    transformers.append(
        (
            "cat",
            categorical_pipeline,
            categorical_features
        )
    )

preprocessor = ColumnTransformer(
    transformers=transformers,
    remainder="drop"
)

# ============================================================
# 17. Define models
# ============================================================

models = {

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        random_state=RANDOM_STATE,
        n_jobs=-1,
        class_weight=None
    ),

    "Logistic Regression": LogisticRegression(
        max_iter=3000,
        random_state=RANDOM_STATE
    ),

    "SVC": SVC(
        kernel="rbf",
        probability=True,
        random_state=RANDOM_STATE
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=7
    ),

    "ANN": MLPClassifier(
        hidden_layer_sizes=(64, 32),
        activation="relu",
        solver="adam",
        alpha=0.001,
        learning_rate_init=0.001,
        max_iter=500,
        early_stopping=True,
        validation_fraction=0.15,
        n_iter_no_change=20,
        random_state=RANDOM_STATE
    )
}

# ============================================================
# 18. Train + evaluate
#
# IMPORTANT:
# RandomOverSampler happens AFTER preprocessing and
# ONLY on the training data.
#
# Test data remains completely untouched.
# ============================================================

results = []

confusion_matrices = {}

predictions = {}

probabilities = {}

trained_pipelines = {}

for model_name, model in models.items():

    print("\n" + "=" * 70)
    print(f"Training: {model_name}")
    print("=" * 70)

    pipeline = ImbPipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),

            (
                "balancer",
                RandomOverSampler(
                    random_state=RANDOM_STATE
                )
            ),

            (
                "model",
                model
            )
        ]
    )

    # Train
    pipeline.fit(X_train, y_train)

    # Predictions
    y_pred = pipeline.predict(X_test)

    # Probabilities for ROC-AUC
    if hasattr(pipeline, "predict_proba"):
        y_prob = pipeline.predict_proba(X_test)[:, 1]

    elif hasattr(pipeline, "decision_function"):
        y_prob = pipeline.decision_function(X_test)

    else:
        y_prob = y_pred

    # Metrics
    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_test,
        y_prob
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "ROC-AUC": roc_auc
    })

    confusion_matrices[model_name] = confusion_matrix(
        y_test,
        y_pred
    )

    predictions[model_name] = y_pred

    probabilities[model_name] = y_prob

    trained_pipelines[model_name] = pipeline

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"ROC-AUC  : {roc_auc:.4f}")

# ============================================================
# 19. Results table
# ============================================================

results_df = pd.DataFrame(results)

metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "ROC-AUC"
]

results_df["Average_Score"] = (
    results_df[metric_columns]
    .mean(axis=1)
)

results_df = results_df.sort_values(
    "Average_Score",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 80)
print("FINAL MODEL COMPARISON")
print("=" * 80)

display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "ROC-AUC": "{:.4f}",
        "Average_Score": "{:.4f}"
    })
)

# ============================================================
# 20. Confusion matrices
# ============================================================

fig, axes = plt.subplots(
    2,
    3,
    figsize=(15, 9)
)

axes = axes.flatten()

for i, model_name in enumerate(models.keys()):

    cm = confusion_matrices[model_name]

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        cbar=False,
        ax=axes[i]
    )

    axes[i].set_title(
        f"{model_name}\nConfusion Matrix"
    )

    axes[i].set_xlabel("Predicted")
    axes[i].set_ylabel("Actual")

# Hide unused subplot
if len(models) < len(axes):
    for j in range(len(models), len(axes)):
        axes[j].axis("off")

plt.tight_layout()
plt.show()

# ============================================================
# 21. Comparative graph - ALL metrics
# ============================================================

plot_df = results_df.set_index("Model")[metric_columns]

ax = plot_df.plot(
    kind="bar",
    figsize=(14, 7)
)

plt.title(
    "Stroke Dataset - Model Comparison Across All Metrics"
)

plt.xlabel("Model")
plt.ylabel("Score")
plt.ylim(0, 1.05)

plt.xticks(rotation=0)
plt.legend(
    title="Metrics",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)

plt.tight_layout()
plt.show()

# ============================================================
# 22. Individual metric graphs
# ============================================================

for metric in metric_columns:

    plt.figure(figsize=(10, 6))

    bars = plt.bar(
        results_df["Model"],
        results_df[metric]
    )

    plt.title(
        f"Stroke Dataset - {metric}"
    )

    plt.xlabel("Model")
    plt.ylabel(metric)

    plt.ylim(0, 1.05)

    plt.xticks(rotation=20)

    for bar, value in zip(
        bars,
        results_df[metric]
    ):
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 0.01,
            f"{value:.3f}",
            ha="center",
            fontsize=9
        )

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()
    plt.show()

# ============================================================
# 23. Average performance
# ============================================================

plt.figure(figsize=(10, 6))

bars = plt.bar(
    results_df["Model"],
    results_df["Average_Score"]
)

plt.title(
    "Stroke Dataset - Average Model Performance"
)

plt.xlabel("Model")
plt.ylabel("Average of Accuracy, Precision, Recall and ROC-AUC")

plt.ylim(0, 1.05)

plt.xticks(rotation=20)

for bar, value in zip(
    bars,
    results_df["Average_Score"]
):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 0.01,
        f"{value:.3f}",
        ha="center",
        fontsize=10
    )

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)

plt.tight_layout()
plt.show()

# ============================================================
# 24. Select best model
# ============================================================

best_model_name = results_df.iloc[0]["Model"]

best_model = trained_pipelines[
    best_model_name
]

print("\n" + "=" * 80)
print("SELECTED MODEL")
print("=" * 80)

print(
    f"Best model based on average of four metrics: "
    f"{best_model_name}"
)

print(
    f"Average score: "
    f"{results_df.iloc[0]['Average_Score']:.4f}"
)

# ============================================================
# 25. Save selected model
# ============================================================

model_filename = (
    "stroke_best_model.pkl"
)

with open(
    model_filename,
    "wb"
) as f:

    pickle.dump(
        best_model,
        f
    )

# Save comparison table
results_filename = (
    "stroke_model_comparison.csv"
)

results_df.to_csv(
    results_filename,
    index=False
)

print("\nFiles saved:")
print(" -", model_filename)
print(" -", results_filename)

# ============================================================
# 26. Final summary
# ============================================================

print("\n" + "=" * 80)
print("STROKE DATASET SUMMARY")
print("=" * 80)

print("Original dataset shape:", df.shape)
print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

print("\nTraining distribution BEFORE balancing:")
print(y_train.value_counts())

print("\nTesting distribution:")
print(y_test.value_counts())

print("\nModel ranking:")
print(
    results_df[
        [
            "Model",
            "Accuracy",
            "Precision",
            "Recall",
            "ROC-AUC",
            "Average_Score"
        ]
    ].to_string(index=False)
)

print(
    f"\nSelected model: {best_model_name}"
)

print(
    "\nIMPORTANT: The test set was never oversampled."
)
```

```python
# ============================================================
# DATASET 4: DIABETES PREDICTION
# Kaggle: mragpavank/diabetes
# ============================================================

# ============================================================
# 1. Install dependencies
# ============================================================

!pip -q install kaggle imbalanced-learn

# ============================================================
# 2. Imports
# ============================================================

import os
import glob
import pickle
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)
from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    roc_auc_score,
    confusion_matrix
)

from imblearn.over_sampling import RandomOverSampler
from imblearn.pipeline import Pipeline as ImbPipeline

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

# ============================================================
# 3. Kaggle authentication
# ============================================================

print("Upload your kaggle.json file:")

uploaded = files.upload()

os.makedirs(
    "/root/.kaggle",
    exist_ok=True
)

# Colab may rename repeated uploads:
# kaggle.json
# kaggle (1).json
# kaggle (2).json
# etc.

kaggle_file = None

for filename in uploaded.keys():

    if filename.lower().endswith(".json"):
        kaggle_file = filename
        break

if kaggle_file is None:

    raise FileNotFoundError(
        "No .json file was uploaded. "
        "Please upload kaggle.json."
    )

with open(
    "/root/.kaggle/kaggle.json",
    "wb"
) as f:

    f.write(
        uploaded[kaggle_file]
    )

os.chmod(
    "/root/.kaggle/kaggle.json",
    0o600
)

print(
    f"Kaggle authentication configured using: "
    f"{kaggle_file}"
)

# ============================================================
# 4. Download EXACT dataset
# ============================================================

DOWNLOAD_DIR = (
    "/content/diabetes_dataset"
)

os.makedirs(
    DOWNLOAD_DIR,
    exist_ok=True
)

print(
    "\nDownloading diabetes dataset..."
)

download_result = os.system(
    "kaggle datasets download "
    "-d mragpavank/diabetes "
    f"-p {DOWNLOAD_DIR} "
    "--unzip"
)

if download_result != 0:

    raise RuntimeError(
        "\nKaggle download failed. "
        "The dataset was not downloaded."
    )

print(
    "Download completed successfully."
)

# ============================================================
# 5. Find CSV files ONLY inside this dataset directory
# ============================================================

csv_files = glob.glob(
    os.path.join(
        DOWNLOAD_DIR,
        "**",
        "*.csv"
    ),
    recursive=True
)

if len(csv_files) == 0:

    raise FileNotFoundError(
        "No CSV file was found inside the "
        "downloaded diabetes dataset."
    )

print(
    "\nCSV files found:"
)

for f in csv_files:
    print(
        " -",
        f
    )

# ============================================================
# 6. Identify the correct CSV
# ============================================================

# We inspect every CSV instead of blindly using csv_files[0].
#
# This prevents accidentally selecting:
# - an old dataset
# - a sample submission
# - a test CSV
# - some unrelated CSV

selected_csv = None

for csv_path in csv_files:

    try:

        temp = pd.read_csv(
            csv_path,
            nrows=10
        )

        print(
            f"\n{os.path.basename(csv_path)}"
        )

        print(
            "Columns:",
            temp.columns.tolist()
        )

        # Candidate target names
        normalized_cols = {
            str(c).strip().lower()
            for c in temp.columns
        }

        target_candidates = {
            "outcome",
            "diabetes",
            "diabetes_binary",
            "diabetes_prediction",
            "diagnosis",
            "target",
            "result",
            "class",
            "label"
        }

        overlap = (
            normalized_cols
            .intersection(
                target_candidates
            )
        )

        if len(overlap) > 0:

            selected_csv = csv_path
            break

    except Exception as e:

        print(
            "Could not inspect:",
            csv_path
        )

# If no obvious target was found,
# use the first CSV as fallback.

if selected_csv is None:

    selected_csv = csv_files[0]

print(
    "\nSelected CSV:"
)

print(
    selected_csv
)

# ============================================================
# 7. Load dataset
# ============================================================

df = pd.read_csv(
    selected_csv
)

print(
    "\nOriginal dataset shape:",
    df.shape
)

print(
    "\nOriginal columns:"
)

print(
    df.columns.tolist()
)

# ============================================================
# 8. Clean column names
# ============================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(
        " ",
        "_",
        regex=False
    )
    .str.replace(
        "/",
        "_",
        regex=False
    )
    .str.replace(
        "-",
        "_",
        regex=False
    )
)

print(
    "\nCleaned columns:"
)

print(
    df.columns.tolist()
)

# ============================================================
# 9. Remove completely empty columns
# ============================================================

df = df.dropna(
    axis=1,
    how="all"
)

# ============================================================
# 10. Remove duplicate rows
# ============================================================

before_duplicates = len(df)

df = (
    df
    .drop_duplicates()
    .reset_index(drop=True)
)

after_duplicates = len(df)

print(
    "\nDuplicate rows removed:",
    before_duplicates - after_duplicates
)

print(
    "Shape after duplicate removal:",
    df.shape
)

# ============================================================
# 11. Detect target column
# ============================================================

target_candidates = [
    "Outcome",
    "outcome",
    "Diabetes",
    "diabetes",
    "Diabetes_Binary",
    "diabetes_binary",
    "Diabetes_Prediction",
    "diabetes_prediction",
    "Diagnosis",
    "diagnosis",
    "Target",
    "target",
    "Result",
    "result",
    "Class",
    "class",
    "Label",
    "label"
]

target_col = None

# Exact candidates first
for candidate in target_candidates:

    if candidate in df.columns:

        target_col = candidate
        break

# Case-insensitive fallback
if target_col is None:

    lower_map = {
        str(col).strip().lower(): col
        for col in df.columns
    }

    for candidate in target_candidates:

        key = candidate.lower()

        if key in lower_map:

            target_col = (
                lower_map[key]
            )

            break

# Last-resort search
if target_col is None:

    for col in df.columns:

        name = (
            str(col)
            .strip()
            .lower()
        )

        if (
            "diabet" in name
            or "outcome" in name
            or "diagnos" in name
        ):

            target_col = col
            break

if target_col is None:

    raise ValueError(
        "\nCould not automatically identify "
        "the target column.\n\n"
        "Available columns:\n"
        f"{df.columns.tolist()}"
    )

print(
    "\nTarget column detected:",
    target_col
)

# ============================================================
# 12. Inspect target
# ============================================================

print(
    "\nOriginal target values:"
)

print(
    df[target_col]
    .value_counts(
        dropna=False
    )
)

# ============================================================
# 13. Clean target
# ============================================================

if (
    df[target_col]
    .dtype == "object"
):

    target_string = (
        df[target_col]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    positive_values = {
        "1",
        "yes",
        "y",
        "true",
        "positive",
        "diabetes",
        "diabetic",
        "present",
        "disease"
    }

    negative_values = {
        "0",
        "no",
        "n",
        "false",
        "negative",
        "no diabetes",
        "non diabetic",
        "non-diabetic",
        "absent",
        "healthy",
        "normal",
        "no disease"
    }

    def encode_target(value):

        if value in positive_values:
            return 1

        if value in negative_values:
            return 0

        return np.nan

    df[target_col] = (
        target_string
        .map(encode_target)
    )

else:

    df[target_col] = pd.to_numeric(
        df[target_col],
        errors="coerce"
    )

# Remove invalid target rows
df = df.dropna(
    subset=[target_col]
).copy()

df[target_col] = (
    df[target_col]
    .astype(int)
)

# Handle possible 1/2 encoding
unique_target = sorted(
    df[target_col]
    .unique()
    .tolist()
)

if unique_target == [1, 2]:

    df[target_col] = (
        df[target_col]
        .map({
            1: 0,
            2: 1
        })
    )

unique_target = sorted(
    df[target_col]
    .unique()
    .tolist()
)

if unique_target != [0, 1]:

    raise ValueError(
        "Target is not binary after cleaning.\n"
        f"Found values: {unique_target}"
    )

print(
    "\nCleaned target distribution:"
)

print(
    df[target_col]
    .value_counts()
)

# ============================================================
# 14. Convert obvious missing-value strings
# ============================================================

missing_strings = [
    "",
    " ",
    "na",
    "n/a",
    "nan",
    "null",
    "none",
    "?",
    "unknown"
]

for col in df.columns:

    if (
        df[col]
        .dtype == "object"
    ):

        df[col] = (
            df[col]
            .astype(str)
            .str.strip()
            .replace(
                missing_strings,
                np.nan
            )
        )

# ============================================================
# 15. Detect numeric columns
# ============================================================

feature_columns = [
    col
    for col in df.columns
    if col != target_col
]

# Attempt numeric conversion for object
# columns where almost all values are numeric.

for col in feature_columns:

    if (
        df[col]
        .dtype == "object"
    ):

        converted = pd.to_numeric(
            df[col],
            errors="coerce"
        )

        non_null_original = (
            df[col]
            .notna()
            .sum()
        )

        non_null_converted = (
            converted
            .notna()
            .sum()
        )

        if (
            non_null_original > 0
            and
            non_null_converted /
            non_null_original >= 0.90
        ):

            df[col] = converted

# ============================================================
# 16. Dataset-specific zero handling
# ============================================================

# Some diabetes datasets use zero for missing
# physiological measurements.
#
# We ONLY apply this to columns that actually exist.
#
# Pregnancies is deliberately NOT included because
# zero pregnancies is valid.

zero_as_missing_candidates = [
    "Glucose",
    "Blood_Glucose",
    "BloodPressure",
    "Blood_Pressure",
    "SkinThickness",
    "Skin_Thickness",
    "Insulin",
    "BMI",
    "Body_Mass_Index"
]

zero_columns = []

for col in df.columns:

    normalized = (
        str(col)
        .strip()
        .lower()
        .replace("_", "")
    )

    for candidate in (
        zero_as_missing_candidates
    ):

        candidate_normalized = (
            candidate
            .lower()
            .replace("_", "")
        )

        if (
            normalized ==
            candidate_normalized
        ):

            zero_columns.append(
                col
            )

            break

zero_columns = list(
    dict.fromkeys(
        zero_columns
    )
)

if zero_columns:

    print(
        "\nConverting zero measurements "
        "to missing values in:"
    )

    print(
        zero_columns
    )

    for col in zero_columns:

        zero_count = (
            df[col] == 0
        ).sum()

        print(
            f"{col}: "
            f"{zero_count} zeros → NaN"
        )

        df.loc[
            df[col] == 0,
            col
        ] = np.nan

else:

    print(
        "\nNo Pima-style zero measurement "
        "columns detected."
    )

# ============================================================
# 17. Missing-value report
# ============================================================

print(
    "\nMissing values:"
)

missing = (
    df.isnull()
    .sum()
    .sort_values(
        ascending=False
    )
)

missing = missing[
    missing > 0
]

if len(missing) == 0:

    print(
        "No missing values."
    )

else:

    print(
        missing
    )

# ============================================================
# 18. Feature engineering
# ============================================================

print(
    "\nPerforming feature engineering..."
)

# -----------------------------
# Helper function
# -----------------------------

def find_column(
    possible_names
):

    normalized_map = {}

    for col in df.columns:

        normalized = (
            str(col)
            .lower()
            .replace("_", "")
            .replace(" ", "")
        )

        normalized_map[
            normalized
        ] = col

    for name in possible_names:

        normalized = (
            name
            .lower()
            .replace("_", "")
            .replace(" ", "")
        )

        if normalized in normalized_map:

            return normalized_map[
                normalized
            ]

    return None


glucose_col = find_column([
    "Glucose",
    "BloodGlucose",
    "GlucoseLevel"
])

bmi_col = find_column([
    "BMI",
    "BodyMassIndex"
])

age_col = find_column([
    "Age"
])

bp_col = find_column([
    "BloodPressure",
    "Blood_Pressure"
])

pregnancy_col = find_column([
    "Pregnancies",
    "Pregnancy"
])

# -----------------------------
# BMI category
# -----------------------------

if bmi_col is not None:

    df["BMI_Category"] = pd.cut(
        df[bmi_col],
        bins=[
            -np.inf,
            18.5,
            25,
            30,
            np.inf
        ],
        labels=[
            "Underweight",
            "Normal",
            "Overweight",
            "Obese"
        ]
    )

# -----------------------------
# Age category
# -----------------------------

if age_col is not None:

    df["Age_Group"] = pd.cut(
        df[age_col],
        bins=[
            -np.inf,
            25,
            40,
            60,
            np.inf
        ],
        labels=[
            "Young",
            "Adult",
            "Middle_Aged",
            "Senior"
        ]
    )

# -----------------------------
# High glucose flag
# -----------------------------

if glucose_col is not None:

    glucose_median = (
        df[glucose_col]
        .median()
    )

    df["High_Glucose_Flag"] = (
        df[glucose_col]
        > glucose_median
    ).astype(int)

# -----------------------------
# BMI × Glucose
# -----------------------------

if (
    bmi_col is not None
    and
    glucose_col is not None
):

    df["BMI_Glucose_Interaction"] = (
        df[bmi_col]
        *
        df[glucose_col]
    )

# -----------------------------
# Age × Glucose
# -----------------------------

if (
    age_col is not None
    and
    glucose_col is not None
):

    df["Age_Glucose_Interaction"] = (
        df[age_col]
        *
        df[glucose_col]
    )

# -----------------------------
# Age × BMI
# -----------------------------

if (
    age_col is not None
    and
    bmi_col is not None
):

    df["Age_BMI_Interaction"] = (
        df[age_col]
        *
        df[bmi_col]
    )

# -----------------------------
# Simple risk-factor count
# -----------------------------

risk_components = []

if glucose_col is not None:

    risk_components.append(
        (
            df[glucose_col]
            >
            df[glucose_col].median()
        ).astype(int)
    )

if bmi_col is not None:

    risk_components.append(
        (
            df[bmi_col]
            >
            df[bmi_col].median()
        ).astype(int)
    )

if age_col is not None:

    risk_components.append(
        (
            df[age_col]
            >
            df[age_col].median()
        ).astype(int)
    )

if bp_col is not None:

    risk_components.append(
        (
            df[bp_col]
            >
            df[bp_col].median()
        ).astype(int)
    )

if len(risk_components) > 0:

    df["Risk_Factor_Count"] = sum(
        risk_components
    )

print(
    "Feature engineering complete."
)

print(
    "Final feature shape:",
    df.shape
)

# ============================================================
# 19. Class balance
# ============================================================

class_counts = (
    df[target_col]
    .value_counts()
    .sort_index()
)

print(
    "\nClass distribution:"
)

for cls, count in (
    class_counts.items()
):

    percentage = (
        count /
        len(df) *
        100
    )

    label = (
        "No Diabetes"
        if cls == 0
        else
        "Diabetes"
    )

    print(
        f"{label}: "
        f"{count} "
        f"({percentage:.2f}%)"
    )

plt.figure(
    figsize=(7, 5)
)

sns.countplot(
    x=target_col,
    data=df
)

plt.title(
    "Diabetes Dataset - Original Class Distribution"
)

plt.xlabel(
    "Diabetes (0 = No, 1 = Yes)"
)

plt.ylabel(
    "Number of Samples"
)

plt.show()

# ============================================================
# 20. Separate X and y
# ============================================================

X = df.drop(
    columns=[target_col]
)

y = df[target_col]

# ============================================================
# 21. Remove obvious ID columns
# ============================================================

id_columns = []

for col in X.columns:

    normalized = (
        str(col)
        .lower()
        .replace("_", "")
        .replace(" ", "")
    )

    if normalized in {
        "id",
        "patientid",
        "patientnumber",
        "recordid",
        "userid"
    }:

        id_columns.append(
            col
        )

if id_columns:

    print(
        "\nRemoving identifier columns:",
        id_columns
    )

    X = X.drop(
        columns=id_columns
    )

# ============================================================
# 22. TRAIN / TEST SPLIT
#
# IMPORTANT:
# Split BEFORE balancing.
# ============================================================

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=RANDOM_STATE,
        stratify=y
    )
)

print(
    "\nTraining shape:",
    X_train.shape
)

print(
    "Testing shape:",
    X_test.shape
)

print(
    "\nTraining distribution BEFORE balancing:"
)

print(
    y_train.value_counts()
)

print(
    "\nTesting distribution:"
)

print(
    y_test.value_counts()
)

# ============================================================
# 23. Determine feature types
# ============================================================

numeric_features = (
    X_train
    .select_dtypes(
        include=[
            "int64",
            "int32",
            "float64",
            "float32"
        ]
    )
    .columns
    .tolist()
)

categorical_features = (
    X_train
    .select_dtypes(
        include=[
            "object",
            "category",
            "bool"
        ]
    )
    .columns
    .tolist()
)

print(
    "\nNumeric features:"
)

print(
    numeric_features
)

print(
    "\nCategorical features:"
)

print(
    categorical_features
)

# ============================================================
# 24. Preprocessing
# ============================================================

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

transformers = []

if numeric_features:

    transformers.append(
        (
            "num",
            numeric_pipeline,
            numeric_features
        )
    )

if categorical_features:

    transformers.append(
        (
            "cat",
            categorical_pipeline,
            categorical_features
        )
    )

preprocessor = ColumnTransformer(
    transformers=transformers,
    remainder="drop"
)

# ============================================================
# 25. Models
# ============================================================

models = {

    "Random Forest":
        RandomForestClassifier(
            n_estimators=300,
            random_state=RANDOM_STATE,
            n_jobs=-1
        ),

    "Logistic Regression":
        LogisticRegression(
            max_iter=3000,
            random_state=RANDOM_STATE
        ),

    "SVC":
        SVC(
            kernel="rbf",
            probability=True,
            random_state=RANDOM_STATE
        ),

    "KNN":
        KNeighborsClassifier(
            n_neighbors=7
        ),

    "ANN":
        MLPClassifier(
            hidden_layer_sizes=(
                64,
                32
            ),
            activation="relu",
            solver="adam",
            alpha=0.001,
            learning_rate_init=0.001,
            max_iter=500,
            early_stopping=True,
            validation_fraction=0.15,
            n_iter_no_change=20,
            random_state=RANDOM_STATE
        )
}

# ============================================================
# 26. Train + evaluate all models
#
# RandomOverSampler is INSIDE the pipeline.
#
# Therefore:
#
# TRAIN:
# preprocessing → balancing → model
#
# TEST:
# preprocessing → model
#
# The test set is NEVER balanced.
# ============================================================

results = []

confusion_matrices = {}

predictions = {}

probabilities = {}

trained_pipelines = {}

for model_name, model in models.items():

    print(
        "\n" + "=" * 70
    )

    print(
        f"Training: {model_name}"
    )

    print(
        "=" * 70
    )

    pipeline = ImbPipeline(
        steps=[

            (
                "preprocessor",
                preprocessor
            ),

            (
                "balancer",
                RandomOverSampler(
                    random_state=RANDOM_STATE
                )
            ),

            (
                "model",
                model
            )
        ]
    )

    # Train
    pipeline.fit(
        X_train,
        y_train
    )

    # Predictions
    y_pred = (
        pipeline
        .predict(X_test)
    )

    # Probability / decision score
    if hasattr(
        pipeline,
        "predict_proba"
    ):

        y_prob = (
            pipeline
            .predict_proba(
                X_test
            )[:, 1]
        )

    elif hasattr(
        pipeline,
        "decision_function"
    ):

        y_prob = (
            pipeline
            .decision_function(
                X_test
            )
        )

    else:

        y_prob = y_pred

    # Metrics
    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_test,
        y_prob
    )

    results.append({

        "Model":
            model_name,

        "Accuracy":
            accuracy,

        "Precision":
            precision,

        "Recall":
            recall,

        "ROC-AUC":
            roc_auc
    })

    confusion_matrices[
        model_name
    ] = confusion_matrix(
        y_test,
        y_pred
    )

    predictions[
        model_name
    ] = y_pred

    probabilities[
        model_name
    ] = y_prob

    trained_pipelines[
        model_name
    ] = pipeline

    print(
        f"Accuracy : {accuracy:.4f}"
    )

    print(
        f"Precision: {precision:.4f}"
    )

    print(
        f"Recall   : {recall:.4f}"
    )

    print(
        f"ROC-AUC  : {roc_auc:.4f}"
    )

# ============================================================
# 27. Model comparison table
# ============================================================

results_df = pd.DataFrame(
    results
)

metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "ROC-AUC"
]

results_df[
    "Average_Score"
] = (
    results_df[
        metric_columns
    ].mean(axis=1)
)

results_df = (
    results_df
    .sort_values(
        "Average_Score",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)

print(
    "\n" + "=" * 80
)

print(
    "FINAL MODEL COMPARISON"
)

print(
    "=" * 80
)

display(
    results_df.style.format({

        "Accuracy":
            "{:.4f}",

        "Precision":
            "{:.4f}",

        "Recall":
            "{:.4f}",

        "ROC-AUC":
            "{:.4f}",

        "Average_Score":
            "{:.4f}"
    })
)

# ============================================================
# 28. Confusion matrices
# ============================================================

fig, axes = plt.subplots(
    2,
    3,
    figsize=(15, 9)
)

axes = axes.flatten()

for i, model_name in enumerate(
    models.keys()
):

    cm = (
        confusion_matrices[
            model_name
        ]
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        cbar=False,
        ax=axes[i]
    )

    axes[i].set_title(
        f"{model_name}\nConfusion Matrix"
    )

    axes[i].set_xlabel(
        "Predicted"
    )

    axes[i].set_ylabel(
        "Actual"
    )

for j in range(
    len(models),
    len(axes)
):

    axes[j].axis(
        "off"
    )

plt.tight_layout()

plt.show()

# ============================================================
# 29. Comparative graph - all metrics
# ============================================================

plot_df = (
    results_df
    .set_index("Model")[
        metric_columns
    ]
)

plot_df.plot(
    kind="bar",
    figsize=(14, 7)
)

plt.title(
    "Diabetes Dataset - Model Comparison"
)

plt.xlabel(
    "Model"
)

plt.ylabel(
    "Score"
)

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=0
)

plt.legend(
    title="Metrics",
    bbox_to_anchor=(
        1.02,
        1
    ),
    loc="upper left"
)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)

plt.tight_layout()

plt.show()

# ============================================================
# 30. Individual metric graphs
# ============================================================

for metric in metric_columns:

    plt.figure(
        figsize=(10, 6)
    )

    bars = plt.bar(
        results_df["Model"],
        results_df[metric]
    )

    plt.title(
        f"Diabetes Dataset - {metric}"
    )

    plt.xlabel(
        "Model"
    )

    plt.ylabel(
        metric
    )

    plt.ylim(
        0,
        1.05
    )

    plt.xticks(
        rotation=20
    )

    for bar, value in zip(
        bars,
        results_df[metric]
    ):

        plt.text(
            bar.get_x()
            + bar.get_width() / 2,
            bar.get_height()
            + 0.01,
            f"{value:.3f}",
            ha="center",
            fontsize=9
        )

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()

    plt.show()

# ============================================================
# 31. Average performance graph
# ============================================================

plt.figure(
    figsize=(10, 6)
)

bars = plt.bar(
    results_df["Model"],
    results_df["Average_Score"]
)

plt.title(
    "Diabetes Dataset - Average Model Performance"
)

plt.xlabel(
    "Model"
)

plt.ylabel(
    "Average of Accuracy, Precision, Recall and ROC-AUC"
)

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=20
)

for bar, value in zip(
    bars,
    results_df["Average_Score"]
):

    plt.text(
        bar.get_x()
        + bar.get_width() / 2,
        bar.get_height()
        + 0.01,
        f"{value:.3f}",
        ha="center",
        fontsize=10
    )

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)

plt.tight_layout()

plt.show()

# ============================================================
# 32. Select best model
# ============================================================

best_model_name = (
    results_df
    .iloc[0]["Model"]
)

best_model = (
    trained_pipelines[
        best_model_name
    ]
)

print(
    "\n" + "=" * 80
)

print(
    "SELECTED MODEL"
)

print(
    "=" * 80
)

print(
    "Best model based on "
    "average of four metrics:",
    best_model_name
)

print(
    "Average score:",
    f"{results_df.iloc[0]['Average_Score']:.4f}"
)

# ============================================================
# 33. Save complete pipeline
# ============================================================

model_filename = (
    "diabetes_best_model.pkl"
)

with open(
    model_filename,
    "wb"
) as f:

    pickle.dump(
        best_model,
        f
    )

results_filename = (
    "diabetes_model_comparison.csv"
)

results_df.to_csv(
    results_filename,
    index=False
)

print(
    "\nFiles saved:"
)

print(
    " -",
    model_filename
)

print(
    " -",
    results_filename
)

# ============================================================
# 34. Final summary
# ============================================================

print(
    "\n" + "=" * 80
)

print(
    "DIABETES DATASET SUMMARY"
)

print(
    "=" * 80
)

print(
    "Final dataset shape:",
    df.shape
)

print(
    "Training samples:",
    len(X_train)
)

print(
    "Testing samples:",
    len(X_test)
)

print(
    "\nTraining distribution BEFORE balancing:"
)

print(
    y_train.value_counts()
)

print(
    "\nTesting distribution:"
)

print(
    y_test.value_counts()
)

print(
    "\nFinal model comparison:"
)

print(
    results_df[
        [
            "Model",
            "Accuracy",
            "Precision",
            "Recall",
            "ROC-AUC",
            "Average_Score"
        ]
    ].to_string(
        index=False
    )
)

print(
    f"\nSelected model: "
    f"{best_model_name}"
)

print(
    "\nIMPORTANT:"
)

print(
    "1. Dataset was split before balancing."
)

print(
    "2. RandomOverSampler was applied "
    "ONLY to training data."
)

print(
    "3. Test data was never oversampled."
)

print(
    "4. Complete preprocessing + "
    "balancing + model pipeline "
    "was saved in the pickle."
)
```

```python
# ============================================================
# ALZHEIMER'S CLINICAL DATASET
# Kaggle: vinodvadde/alzeimers-clinical-dataset
# ============================================================

!pip -q install kaggle imbalanced-learn

import os
import glob
import shutil
import pickle
import warnings
import subprocess
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    LabelEncoder
)
from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    roc_auc_score,
    confusion_matrix
)

from imblearn.over_sampling import RandomOverSampler
from imblearn.pipeline import Pipeline as ImbPipeline

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

# ============================================================
# 1. KAGGLE AUTHENTICATION
# ============================================================

print("=" * 70)
print("STEP 1: KAGGLE AUTHENTICATION")
print("=" * 70)

uploaded = files.upload()

os.makedirs("/root/.kaggle", exist_ok=True)

kaggle_file = None

for filename in uploaded.keys():
    if filename.lower().endswith(".json"):
        kaggle_file = filename
        break

if kaggle_file is None:
    raise FileNotFoundError(
        "No .json file was uploaded. Please upload your Kaggle API key."
    )

with open("/root/.kaggle/kaggle.json", "wb") as f:
    f.write(uploaded[kaggle_file])

os.chmod("/root/.kaggle/kaggle.json", 0o600)

print(f"✓ Kaggle credentials loaded from: {kaggle_file}")


# ============================================================
# 2. DOWNLOAD EXACT ALZHEIMER DATASET
# ============================================================

print("\n" + "=" * 70)
print("STEP 2: DOWNLOADING ALZHEIMER'S DATASET")
print("=" * 70)

DATASET_DIR = "/content/alzheimers_dataset"

if os.path.exists(DATASET_DIR):
    shutil.rmtree(DATASET_DIR)

os.makedirs(DATASET_DIR, exist_ok=True)

download_cmd = [
    "kaggle",
    "datasets",
    "download",
    "-d",
    "vinodvadde/alzeimers-clinical-dataset",
    "-p",
    DATASET_DIR,
    "--unzip"
]

result = subprocess.run(
    download_cmd,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError(
        "Kaggle dataset download failed. Check the Kaggle credentials."
    )

print("✓ Dataset downloaded successfully.")


# ============================================================
# 3. FIND CSV ONLY INSIDE THIS DATASET DIRECTORY
# ============================================================

csv_files = glob.glob(
    os.path.join(DATASET_DIR, "**", "*.csv"),
    recursive=True
)

if not csv_files:
    raise FileNotFoundError(
        "No CSV file found inside the Alzheimer's dataset."
    )

print("\nCSV files discovered:")

for i, f in enumerate(csv_files):
    print(f"  [{i}] {f}")

# Prefer the largest CSV if multiple files exist
csv_file = max(csv_files, key=os.path.getsize)

print(f"\n✓ Selected CSV: {csv_file}")


# ============================================================
# 4. LOAD DATA
# ============================================================

df = pd.read_csv(csv_file)

print("\n" + "=" * 70)
print("STEP 3: DATASET INSPECTION")
print("=" * 70)

print("Shape:", df.shape)

print("\nColumns:")
for col in df.columns:
    print(" ", repr(col))

print("\nFirst 5 rows:")
display(df.head())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())


# ============================================================
# 5. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(r"\s+", "_", regex=True)
    .str.replace(r"[/\-]+", "_", regex=True)
    .str.replace(r"[()]", "", regex=True)
)

# Remove completely empty columns
df = df.dropna(axis=1, how="all")

# Remove duplicate rows
before_duplicates = len(df)
df = df.drop_duplicates().reset_index(drop=True)

print("\nDuplicates removed:", before_duplicates - len(df))
print("Shape after cleaning:", df.shape)

print("\nCleaned columns:")
print(list(df.columns))


# ============================================================
# 6. AUTOMATIC TARGET DETECTION
# ============================================================

print("\n" + "=" * 70)
print("STEP 4: TARGET DETECTION")
print("=" * 70)

target_candidates = [
    "Diagnosis",
    "diagnosis",
    "Diagnosis_Status",
    "Diagnosis_status",
    "Alzheimer_Diagnosis",
    "Alzheimer_s_Disease",
    "Alzheimers_Disease",
    "Alzheimer",
    "Alzheimers",
    "Dementia",
    "dementia",
    "Dementia_Status",
    "Dementia_status",
    "Class",
    "class",
    "Target",
    "target",
    "Outcome",
    "outcome",
    "Label",
    "label",
    "Disease",
    "disease",
    "Result",
    "result"
]

target_col = None

# Exact candidate matching
for candidate in target_candidates:
    if candidate in df.columns:
        target_col = candidate
        break

# Keyword fallback
if target_col is None:
    keywords = [
        "diagnos",
        "dementia",
        "alzheimer",
        "target",
        "outcome",
        "class",
        "label"
    ]

    for col in df.columns:
        lower_col = col.lower()

        if any(keyword in lower_col for keyword in keywords):
            target_col = col
            break

if target_col is None:
    raise ValueError(
        "Could not automatically identify the target column.\n"
        f"Available columns:\n{list(df.columns)}"
    )

print(f"✓ Target column detected: {target_col}")

print("\nTarget value distribution BEFORE encoding:")
print(df[target_col].value_counts(dropna=False))


# ============================================================
# 7. TARGET ENCODING
# ============================================================

print("\n" + "=" * 70)
print("STEP 5: TARGET ENCODING")
print("=" * 70)

y_raw = df[target_col].copy()

# Clean string targets
if y_raw.dtype == "object" or str(y_raw.dtype).startswith("category"):

    y_clean = (
        y_raw
        .astype(str)
        .str.strip()
        .str.lower()
    )

    positive_words = {
        "1",
        "yes",
        "y",
        "true",
        "positive",
        "present",
        "disease",
        "dementia",
        "alzheimer",
        "alzheimers",
        "alzheimer's",
        "diagnosed",
        "patient",
        "affected"
    }

    negative_words = {
        "0",
        "no",
        "n",
        "false",
        "negative",
        "absent",
        "healthy",
        "normal",
        "control",
        "none",
        "not diagnosed",
        "non-demented",
        "nondemented"
    }

    encoded = []

    for value in y_clean:

        if value in positive_words:
            encoded.append(1)

        elif value in negative_words:
            encoded.append(0)

        else:
            encoded.append(np.nan)

    encoded = pd.Series(encoded, index=df.index)

    # If some values weren't recognized, use LabelEncoder
    if encoded.notna().sum() != len(encoded):

        print(
            "\nSome target values were not recognized automatically."
        )
        print("Using LabelEncoder instead.")

        valid_mask = y_raw.notna()

        le = LabelEncoder()
        encoded_values = le.fit_transform(
            y_raw[valid_mask].astype(str)
        )

        y = pd.Series(
            np.nan,
            index=df.index,
            dtype=float
        )

        y.loc[valid_mask] = encoded_values

        print(
            "Label encoding:",
            dict(
                zip(
                    le.classes_,
                    le.transform(le.classes_)
                )
            )
        )

    else:
        y = encoded

else:

    y_numeric = pd.to_numeric(
        y_raw,
        errors="coerce"
    )

    unique_values = sorted(
        y_numeric.dropna().unique()
    )

    print("Numeric target values:", unique_values)

    # Handle common 1/2 encoding
    if set(unique_values).issubset({1, 2}):
        y = y_numeric.map({
            1: 0,
            2: 1
        })

    elif set(unique_values).issubset({0, 1}):
        y = y_numeric

    else:
        # General binary fallback
        if len(unique_values) != 2:
            raise ValueError(
                f"Target must be binary for this pipeline. "
                f"Found values: {unique_values}"
            )

        mapping = {
            unique_values[0]: 0,
            unique_values[1]: 1
        }

        y = y_numeric.map(mapping)

# Remove rows with invalid target
valid_target = y.notna()

df = df.loc[valid_target].reset_index(drop=True)
y = y.loc[valid_target].astype(int).reset_index(drop=True)

print("\nEncoded target distribution:")
print(y.value_counts())

print("\nTarget mapping:")
print("0 = negative/control class")
print("1 = positive/disease class")


# ============================================================
# 8. PREPARE FEATURES
# ============================================================

X = df.drop(columns=[target_col]).copy()


# ============================================================
# 9. REMOVE OBVIOUS ID COLUMNS
# ============================================================

id_like_columns = []

for col in X.columns:

    lower = col.lower()

    if (
        lower in {
            "id",
            "patient_id",
            "patientid",
            "record_id",
            "recordid",
            "unnamed_0"
        }
        or lower.startswith("unnamed")
    ):
        id_like_columns.append(col)

if id_like_columns:

    print("\nRemoving ID-like columns:")
    print(id_like_columns)

    X = X.drop(columns=id_like_columns)


# ============================================================
# 10. CLEAN STRING VALUES
# ============================================================

for col in X.select_dtypes(include=["object"]).columns:

    X[col] = (
        X[col]
        .astype(str)
        .str.strip()
        .replace({
            "": np.nan,
            "nan": np.nan,
            "None": np.nan,
            "none": np.nan,
            "NA": np.nan,
            "N/A": np.nan,
            "?": np.nan
        })
    )


# ============================================================
# 11. CONVERT NUMERIC-LIKE OBJECT COLUMNS
# ============================================================

for col in X.columns:

    if X[col].dtype == "object":

        converted = pd.to_numeric(
            X[col],
            errors="coerce"
        )

        valid_ratio = converted.notna().mean()

        if valid_ratio >= 0.90:
            X[col] = converted


# ============================================================
# 12. ALZHEIMER/CLINICAL FEATURE ENGINEERING
# ============================================================

print("\n" + "=" * 70)
print("STEP 6: FEATURE ENGINEERING")
print("=" * 70)

created_features = []


def find_column(possible_names):

    for name in possible_names:

        if name in X.columns:
            return name

    # Case-insensitive fallback
    lower_map = {
        col.lower(): col
        for col in X.columns
    }

    for name in possible_names:

        if name.lower() in lower_map:
            return lower_map[name.lower()]

    return None


age_col = find_column([
    "Age",
    "age",
    "Age_Years"
])

mmse_col = find_column([
    "MMSE",
    "MMSE_Score",
    "MMSE_Score_"
])

cdr_col = find_column([
    "CDR",
    "CDR_Score",
    "Clinical_Dementia_Rating"
])

education_col = find_column([
    "EDUC",
    "Education",
    "Education_Years",
    "Years_of_Education"
])

ses_col = find_column([
    "SES",
    "Socioeconomic_Status"
])


# Age-related features
if age_col is not None:

    X[age_col] = pd.to_numeric(
        X[age_col],
        errors="coerce"
    )

    X["Age_Squared"] = X[age_col] ** 2
    created_features.append("Age_Squared")

    X["Age_Group"] = pd.cut(
        X[age_col],
        bins=[0, 50, 60, 70, 80, 200],
        labels=[
            "Below_50",
            "50_60",
            "60_70",
            "70_80",
            "80_plus"
        ],
        include_lowest=True
    )

    created_features.append("Age_Group")


# MMSE features
if mmse_col is not None:

    X[mmse_col] = pd.to_numeric(
        X[mmse_col],
        errors="coerce"
    )

    X["Low_MMSE_Flag"] = (
        X[mmse_col] < 24
    ).astype(int)

    created_features.append("Low_MMSE_Flag")


# CDR features
if cdr_col is not None:

    X[cdr_col] = pd.to_numeric(
        X[cdr_col],
        errors="coerce"
    )

    X["Elevated_CDR_Flag"] = (
        X[cdr_col] > 0
    ).astype(int)

    created_features.append("Elevated_CDR_Flag")


# Education feature
if education_col is not None:

    X[education_col] = pd.to_numeric(
        X[education_col],
        errors="coerce"
    )

    X["Education_Low_Flag"] = (
        X[education_col] <
        X[education_col].median()
    ).astype(int)

    created_features.append("Education_Low_Flag")


# Interaction: Age × MMSE
if age_col is not None and mmse_col is not None:

    X["Age_MMSE_Interaction"] = (
        X[age_col] * X[mmse_col]
    )

    created_features.append(
        "Age_MMSE_Interaction"
    )


# Interaction: Age × CDR
if age_col is not None and cdr_col is not None:

    X["Age_CDR_Interaction"] = (
        X[age_col] * X[cdr_col]
    )

    created_features.append(
        "Age_CDR_Interaction"
    )


# Interaction: MMSE × CDR
if mmse_col is not None and cdr_col is not None:

    X["MMSE_CDR_Interaction"] = (
        X[mmse_col] * X[cdr_col]
    )

    created_features.append(
        "MMSE_CDR_Interaction"
    )


print("Created features:")

if created_features:
    for feature in created_features:
        print(" ✓", feature)
else:
    print("No dataset-specific engineered features were created.")
    print("The original clinical variables will still be used.")


# ============================================================
# 13. REMOVE CONSTANT COLUMNS
# ============================================================

constant_columns = [
    col for col in X.columns
    if X[col].nunique(dropna=False) <= 1
]

if constant_columns:

    print("\nRemoving constant columns:")
    print(constant_columns)

    X = X.drop(columns=constant_columns)


# ============================================================
# 14. CLASS BALANCE CHECK
# ============================================================

print("\n" + "=" * 70)
print("STEP 7: CLASS BALANCE")
print("=" * 70)

class_counts = y.value_counts().sort_index()

print(class_counts)

plt.figure(figsize=(7, 5))

plt.bar(
    ["Class 0", "Class 1"],
    [
        class_counts.get(0, 0),
        class_counts.get(1, 0)
    ]
)

plt.title("Alzheimer's Dataset - Class Distribution")
plt.ylabel("Number of Samples")

for i, value in enumerate([
    class_counts.get(0, 0),
    class_counts.get(1, 0)
]):
    plt.text(
        i,
        value,
        str(value),
        ha="center",
        va="bottom"
    )

plt.show()


# ============================================================
# 15. TRAIN / TEST SPLIT
# IMPORTANT:
# SPLIT FIRST, BALANCE ONLY TRAINING DATA
# ============================================================

print("\n" + "=" * 70)
print("STEP 8: TRAIN / TEST SPLIT")
print("=" * 70)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

print("\nOriginal training distribution:")
print(y_train.value_counts())

print("\nOriginal testing distribution:")
print(y_test.value_counts())


# ============================================================
# 16. IDENTIFY NUMERIC / CATEGORICAL FEATURES
# ============================================================

numeric_features = X_train.select_dtypes(
    include=["int64", "int32", "float64", "float32"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nNumeric features:", numeric_features)
print("\nCategorical features:", categorical_features)


# ============================================================
# 17. PREPROCESSING
# ============================================================

numeric_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    )
])

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "onehot",
        OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    )
])

transformers = []

if numeric_features:
    transformers.append(
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        )
    )

if categorical_features:
    transformers.append(
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    )

preprocessor = ColumnTransformer(
    transformers=transformers,
    remainder="drop"
)


# ============================================================
# 18. DEFINE FIVE MODELS
# ============================================================

models = {

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        random_state=RANDOM_STATE,
        class_weight=None,
        n_jobs=-1
    ),

    "Logistic Regression": LogisticRegression(
        max_iter=3000,
        random_state=RANDOM_STATE
    ),

    "SVC": SVC(
        kernel="rbf",
        probability=True,
        random_state=RANDOM_STATE
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=7
    ),

    "ANN": MLPClassifier(
        hidden_layer_sizes=(64, 32),
        activation="relu",
        solver="adam",
        max_iter=1000,
        early_stopping=True,
        validation_fraction=0.15,
        random_state=RANDOM_STATE
    )
}


# ============================================================
# 19. TRAIN + EVALUATE ALL MODELS
# ============================================================

print("\n" + "=" * 70)
print("STEP 9: TRAINING FIVE MODELS")
print("=" * 70)

results = []
trained_pipelines = {}
confusion_matrices = {}

for model_name, model in models.items():

    print(f"\n{'-' * 60}")
    print(f"Training: {model_name}")
    print(f"{'-' * 60}")

    # IMPORTANT:
    # RandomOverSampler is INSIDE the pipeline.
    # Therefore it only touches the training data.
    pipeline = ImbPipeline([
        (
            "preprocessor",
            preprocessor
        ),
        (
            "balancer",
            RandomOverSampler(
                random_state=RANDOM_STATE
            )
        ),
        (
            "model",
            model
        )
    ])

    pipeline.fit(
        X_train,
        y_train
    )

    y_pred = pipeline.predict(X_test)

    # ROC-AUC
    if hasattr(
        pipeline.named_steps["model"],
        "predict_proba"
    ):
        y_prob = pipeline.predict_proba(
            X_test
        )[:, 1]

    else:
        y_scores = pipeline.decision_function(
            X_test
        )

        y_prob = y_scores

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    try:
        roc_auc = roc_auc_score(
            y_test,
            y_prob
        )
    except:
        roc_auc = np.nan

    average_score = np.nanmean([
        accuracy,
        precision,
        recall,
        roc_auc
    ])

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "ROC-AUC": roc_auc,
        "Average": average_score
    })

    trained_pipelines[model_name] = pipeline

    confusion_matrices[model_name] = confusion_matrix(
        y_test,
        y_pred
    )

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"ROC-AUC  : {roc_auc:.4f}")
    print(f"Average  : {average_score:.4f}")


# ============================================================
# 20. RESULTS TABLE
# ============================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Average",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("STEP 10: MODEL COMPARISON")
print("=" * 70)

display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "ROC-AUC": "{:.4f}",
        "Average": "{:.4f}"
    })
)


# ============================================================
# 21. CONFUSION MATRICES
# ============================================================

print("\n" + "=" * 70)
print("STEP 11: CONFUSION MATRICES")
print("=" * 70)

for model_name in models.keys():

    plt.figure(figsize=(5, 4))

    sns.heatmap(
        confusion_matrices[model_name],
        annot=True,
        fmt="d",
        cmap="Blues",
        cbar=False,
        xticklabels=[
            "Predicted 0",
            "Predicted 1"
        ],
        yticklabels=[
            "Actual 0",
            "Actual 1"
        ]
    )

    plt.title(
        f"{model_name} - Confusion Matrix"
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.tight_layout()
    plt.show()


# ============================================================
# 22. COMPARATIVE GRAPH - ALL METRICS
# ============================================================

print("\n" + "=" * 70)
print("STEP 12: COMPARATIVE ANALYSIS")
print("=" * 70)

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "ROC-AUC"
]

plot_df = results_df.set_index(
    "Model"
)[metrics]

ax = plot_df.plot(
    kind="bar",
    figsize=(13, 6)
)

plt.title(
    "Alzheimer's Dataset - Model Performance Comparison"
)

plt.ylabel("Score")
plt.ylim(0, 1.05)
plt.xticks(rotation=25)
plt.legend(
    title="Metrics",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()


# ============================================================
# 23. INDIVIDUAL METRIC COMPARISON
# ============================================================

for metric in metrics:

    plt.figure(figsize=(10, 5))

    plt.bar(
        results_df["Model"],
        results_df[metric]
    )

    plt.title(
        f"{metric} Comparison"
    )

    plt.ylabel(metric)
    plt.ylim(0, 1.05)

    plt.xticks(rotation=25)

    for i, value in enumerate(
        results_df[metric]
    ):
        plt.text(
            i,
            value,
            f"{value:.3f}",
            ha="center",
            va="bottom"
        )

    plt.tight_layout()
    plt.show()


# ============================================================
# 24. AVERAGE PERFORMANCE
# ============================================================

plt.figure(figsize=(10, 5))

plt.bar(
    results_df["Model"],
    results_df["Average"]
)

plt.title(
    "Average Performance Across Accuracy, Precision, Recall and ROC-AUC"
)

plt.ylabel("Average Score")
plt.ylim(0, 1.05)

plt.xticks(rotation=25)

for i, value in enumerate(
    results_df["Average"]
):
    plt.text(
        i,
        value,
        f"{value:.3f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()


# ============================================================
# 25. SELECT BEST MODEL
# ============================================================

best_model_name = results_df.iloc[0]["Model"]

best_pipeline = trained_pipelines[
    best_model_name
]

best_average = results_df.iloc[0]["Average"]

print("\n" + "=" * 70)
print("STEP 13: FINAL MODEL SELECTION")
print("=" * 70)

print(f"Selected model: {best_model_name}")
print(f"Average score : {best_average:.4f}")

print("\nReason:")
print(
    "The model was selected using the mean of "
    "Accuracy, Precision, Recall and ROC-AUC."
)

print(
    "\nThis avoids selecting a model purely because "
    "of raw accuracy on an imbalanced clinical dataset."
)


# ============================================================
# 26. SAVE COMPLETE PIPELINE
# ============================================================

MODEL_PATH = "/content/alzheimers_best_model.pkl"
RESULTS_PATH = "/content/alzheimers_model_comparison.csv"

with open(
    MODEL_PATH,
    "wb"
) as f:

    pickle.dump(
        best_pipeline,
        f
    )

results_df.to_csv(
    RESULTS_PATH,
    index=False
)

print("\n" + "=" * 70)
print("STEP 14: SAVED FILES")
print("=" * 70)

print(f"✓ Complete model pipeline: {MODEL_PATH}")
print(f"✓ Model comparison CSV    : {RESULTS_PATH}")


# ============================================================
# 27. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print(f"Dataset shape after cleaning: {df.shape}")
print(f"Feature count: {X.shape[1]}")
print(f"Training samples: {len(X_train)}")
print(f"Testing samples: {len(X_test)}")

print("\nModels trained:")
for model_name in models:
    print(" ✓", model_name)

print("\nBest model:")
print(f" ★ {best_model_name}")

print("\nBest average score:")
print(f" ★ {best_average:.4f}")

print("\nTest-set class distribution was kept untouched.")
print("Random oversampling was applied ONLY inside the training pipeline.")

print("\n" + "=" * 70)
print("ALZHEIMER'S DATASET COMPLETE")
print("=" * 70)
```
